```python
#task: 01, 2, 3

import pandas as pd

data = {
    "Students": ["Ali", "Sara", "Ahmed" , "Ayesha", "Zain"],
   "City": ["Karachi", "Lahore", "Multan", "Peshawar", "Karachi"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)
print(df)

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nTotal Students:")
print(df["Students"])

print("\nMaximum Marks:")
print(df["Marks"].max())

print("\nMinimum Marks:")
print(df["Marks"].min())

print("\nPassed Students:")
passed = df[df["Marks"] >= 50]
print(passed)

print("\nFailed Students:")
failed = df[df["Marks"] < 50]
print(failed)

print("\nStudents from Karachi:")
students_from_karachi = df[df["City"] == "Karachi"]
print(students_from_karachi)
```

```python
#task: 05

import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [80, np.nan, 70, np.nan, 90]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Number of missing values
print("\nNumber of missing values:")
print(df["Marks"].isnull().sum())

# Calculate average marks
average_marks = df["Marks"].mean()

# Fill missing values with average marks
df["Marks"] = df["Marks"].fillna(average_marks)


print("\nCleaned DataFrame:")
print(df)
```

```python
#handling missing values


data = {
    "Name": ["Ayan", "Amna", "Mariam"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Check missing values
print("\nMissing values:")
print(df.isnull())

# Count missing values
print("\nNumber of missing values:")
print(df.isnull().sum())

# Fill missing marks with average
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Display cleaned DataFrame
print("\nCleaned DataFrame:")
print(df)

```

```python
#handling duplicates

data = {
    "Name": ["Ayan", "Amna", "Mariam"],
    "Marks": [80, 90, 80]
}

df = pd.DataFrame(data)

print(df.duplicated())

#count duplicates
print(df.duplicated().sum())

#to remove duplicates
df = df.drop_duplicates()

print(df)
```

```python
#sorting data

data = {
    "Name": ["Ayan", "Amna", "Mariam"],
    "Marks": [70, 95, 80]
}

df = pd.DataFrame(data)

#in ascending
print("\nMarks in Ascending order:")
print(df.sort_values("Marks"))

#in descending order
print("\nMarks in Descending order:")
print(df.sort_values("Marks", ascending=False))
```

```python
#aggregation

marks = df["Marks"]
print(marks.sum())
print(marks.mean())
print(marks.max())
print(marks.min())
print(marks.count())
```

```python
#groupby

data = {
    "Name": ["Ayan", "Amna", "Mariam", "Maham"],
    "Course": ["AI", "Web", "AI", "Web"],
    "Marks": [80, 90, 70, 85]
}

df = pd.DataFrame(data)
result = df.groupby("Course")["Marks"].mean()

print(result)
```

```python
#combine data frames

df1 = pd.DataFrame({
    "Name": ["Ayan", "Amna"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Mariam", "Maham"],
    "Marks": [75, 85]
})
final_df = pd.concat([df1, df2], ignore_index=True)

print(final_df)
```

```python
#merge

students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ayan", "Mariam", "Amna"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})
final_df = pd.merge(students, marks, on="ID")

print(final_df)
```
